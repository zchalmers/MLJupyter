# Webcam Speed Limit Demo

This notebook loads the curated trained detector from `models/speed_limit_detector_best.pt` and runs live webcam inference.


```python
from pathlib import Path
import cv2
from ultralytics import YOLO

def resolve_model_path() -> Path:
    candidates = [
        Path("models/speed_limit_detector_best.pt"),
        Path("../models/speed_limit_detector_best.pt"),
    ]
    for candidate in candidates:
        path = candidate.resolve()
        if path.exists():
            return path
    raise FileNotFoundError(
        "Could not find the curated model weights. Train the detector first or place the weight file in models/."
    )

MODEL_PATH = resolve_model_path()
model = YOLO(str(MODEL_PATH))
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    raise RuntimeError("Unable to access the default webcam.")

try:
    while True:
        success, frame = cap.read()
        if not success:
            raise RuntimeError("Failed to read a frame from the webcam.")

        results = model(frame, verbose=False)
        annotated_frame = results[0].plot()
        cv2.imshow("Speed Limit Detector", annotated_frame)

        if cv2.waitKey(1) & 0xFF == ord("q"):
            break
finally:
    cap.release()
    cv2.destroyAllWindows()
```
